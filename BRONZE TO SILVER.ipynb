```python
import pandas as pd
from pyspark.sql import SparkSession

# Load the data from the table into a Spark DataFrame
df_spark = spark.table("`data-1`.`example-1`.`users`")

# Convert to pandas DataFrame
df = df_spark.toPandas()

display(df.head())
```

```python
# Fix the date column - replace '.' with '/' separator
df['signup_date']=df['signup_date'].str.replace('.', '/',regex=False)


#  convert signup_date from string to datetime
df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')

# Remove duplicates based on user_id (keeping first occurrence)
df_cleaned = df.drop_duplicates(subset='user_id', keep='first')
```

```python
# Display the cleaned data
display(df_cleaned)
```

```python
# Convert back to Spark DataFrame
df_spark_cleaned = spark.createDataFrame(df_cleaned)

# Save to cleaned_silver table
df_spark_cleaned.write.mode("overwrite").saveAsTable("`data-1`.`example-1`.`cleaned_silver`")
```
