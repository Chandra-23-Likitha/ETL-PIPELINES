```python
from pyspark import pipelines as dp
from pyspark.sql import functions as F

@dp.materialized_view(
    name="cleanedsilver",
    comment="Cleaned user data with fixed date column and duplicates removed"
)
def cleanedsilver():
    # Load the data from the bronze table
    df = spark.read.table("`data-1`.`example-1`.`users`")
    
    # Fix the date column - convert signup_date from string to date
    df = df.withColumn("signup_date", F.to_date(F.col("signup_date"), "M/d/yy"))
    
    # Remove duplicates based on user_id (keeping first occurrence)
    df = df.dropDuplicates(["user_id"])
    
    return df
```
